```python
colonnes = ["annee_naissance", "age", "metro_ens", "metro_h", "metro_f",
            "france_ens", "france_h", "france_f"]

pop = pd.read_excel(chemin, sheet_name="2026", header=None,
                    skiprows=6, nrows=106, usecols="A:H", names=colonnes)
print(pop.shape)
pop.tail(3)
```

```python
print(pop["france_ens"].sum())
print(pop["france_h"].sum(), pop["france_f"].sum())
print((pop["france_h"] + pop["france_f"] - pop["france_ens"]).abs().max())
```

```python
N = np.vstack([pop["france_h"].to_numpy(), pop["france_f"].to_numpy()]).astype("float64")
print(N.shape, N.dtype)
```

```python
N.sum()
```

```python
j = np.array([[1, 2, 3],
              [4, 5, 6]])
print(j.sum(axis=0))   # [5 7 9]  -> 3 valeurs : une par âge
print(j.sum(axis=1))   # [ 6 15]  -> 2 valeurs : une par sexe
```

```python
total = N.sum()
par_sexe = N.sum(axis=1)
par_age = N.sum(axis=0)

print(total)
print(par_sexe)
print(par_age.shape)
print(par_age[0], par_age[105])
```
